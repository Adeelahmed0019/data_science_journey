## 1. Characteristics of a Tuple

####  A tuple is an `ordered`, `immutable`(unchangable) collection of elements `enclosed` in `round` parentheses `()`.

#### • **Ordered:** Items maintain their `exact` creation `position` (index).
#### • **Immutable:** You `cannot` *change*, *add*, or *remove* elements `once` the tuple is created.
#### • **Allows Duplicates:** You can `repeat` the `same values` multiple times.
#### • **Heterogeneous:** It can `hold` a `mix` of different `data types` (strings, integers, lists, etc.) simultaneously.


```python
tuple_a = (1,2,"Hi", 3.3, True, None, False, [1,2,3], (4,5,6), {"key": "value"},2,2)
print("The type of tuple_a is:", type(tuple_a))
print("The tuple is:", tuple_a)
```

#### Immutable

```python
point = (10, 20)
point[0] = 30  # This will raise a TypeError because tuples are immutable
```

### **2. The Single-Element Trap** 
#### If you want to `create a tuple` with `only` one item, *writing (5)* will `not work.`
#### Python will `evaluate` the parentheses `as standard math brackets` and treat the `variable as an integer.`
#### The Fix: You must include a **trailing comma**


```python
wrong_tuple = (5)     # type(wrong_tuple) is <class 'int'>
print("The type of wrong_tuple is:", type(wrong_tuple))
print("The value of wrong_tuple is:", wrong_tuple)

right_tuple = (5,)    # type(right_tuple) is <class 'tuple'>
print("The type of right_tuple is:", type(right_tuple))
print("The value of right_tuple is:", right_tuple)
```

### 3. **When to Use a Tuple over a List?**

#### • **Data Integrity:** Use tuples when you `want to protect` your data from accidental `modifications` (e.g., geographic coordinates (latitude, longitude), configuration settings).
#### • **Performance:** Tuples are `slightly faster` and consume less memory than lists because they are static.
#### • ***Dictionary Keys:** Because tuples are `immutable`, they are **"hashable."** This means you can `use a tuple as a key in a dictionary;` you cannot do this with a list.


### 4. **Tuple Packing and Unpacking**
#### • **Packing:** Gathering `multiple individual values` *together* into a `single tuple.`
#### • **Unpacking:** `Extracting` the values `inside a tuple` back out `into separate individual variables.`

# Packing

```python
user_data = "Alice", 28, "Data scienticst" 
print(type(user_data))
print(user_data)  # Output: ('Alice', 28, 'Data scientist')
```

# Unpacking

```python
name, age, job = user_data
print(name)  # Output: Alice
print(age)   # Output: 28
print(job)   # Output: Data scientist
```

### 5. **Classic Swap of Two Variables Without a Temp Variable**
#### In many programming languages, `swapping variables` requires a `temporary` storage variable. 
#### Python `leverages` tuple `packing and unpacking` to handle this in one clean line.

```python
a , b = 1, 2
print("The value of a is:", a)  # Output: The value of a is: 1
print("The value of b is:", b)  # Output: The value of b is: 2
a , b = b , a
print("After swapping:")
print("The value of `a` is", a)  # Output: a: 2
print("The value of `b` is ", b)  # Output: b: 1
```

### 6. **Star Unpacking (*)**
#### If you want to `unpack` a tuple but only care about extracting a few specific elements, use the star * operator to bundle the remaining elements into a list.

```python
numbers = (1, 2, 3, 4, 5)

# Grab the first, the last, and group everything in the middle
first, *middle, last = numbers
print(first)   # Output: 1
print(type(first))
print(middle)  # Output: [2, 3, 4]  (Notice it unpacks into a list!)
print(type(middle))
print(last)    # Output: 5
print(type(last))
```

### 7. **Built-in Functions for Tuples**
#### These global Python functions accept a tuple as their target:


#### 1. len() - Count total elements

```python
scores = (45, 82, 19, 94, 55, 19)
total_items = len(scores)
print("Total number of items:", total_items)  # Output: 6
```

#### 2. max() & 3. min() - Find extreme values

```python
highest = max(scores)
lowest = min(scores)
print(f"Highest score: {highest}, Lowest score: {lowest}")  # Output: Highest score: 94, Lowest score: 19
```

#### 4. sum() - Add numerical values together

```python
total_sum = sum(scores)
print("Sum of all numbers:", total_sum)  # Output: 314
```

### 5. sorted() - Sort items 
#### CRITICAL NOTE: Because tuples are immutable, sorted() returns a BRAND NEW sorted LIST.
#### It does not alter the original tuple.

```python
sorted_list = sorted(scores)
sorted_tuple = tuple(sorted_list)  # Convert it back to a tuple if needed
print("Sorted list:", sorted_list)    # Output: [19, 19, 45, 55, 82, 94]
print("Output of sorted():", sorted_list)     # Output: [19, 19, 45, 55, 82, 94] (a list!)
print("Converted back to tuple:", sorted_tuple) # Output: (19, 19, 45, 55, 82, 94)
```

#### 6. tuple() - Type conversion

```python
list_data = ["Python", "Git", "VSCode"]
converted_tuple = tuple(list_data)
print("Converted from list to tuple:", converted_tuple)  # Output: ('Python', 'Git', 'VSCode')
```

#### 7. all() - Returns True if ALL elements evaluate to True

```python
all_true = (True, 1, "Hello")  # Non-empty strings and non-zero numbers evaluate to True
mixed_booleans = (True, False, True)
print("all(mixed_booleans):", all(mixed_booleans))  # Output: False (because one item is False)
print("all(all_true):", all(all_true))              # Output: True
```

#### 8. any() - Returns True if AT LEAST ONE element is True


```python
all_false = (False, 0, "")     # 0 and empty strings evaluate to False
print("any(mixed_booleans):", any(mixed_booleans))  # Output: True (because at least one is True)
print("any(all_false):", any(all_false))            # Output: False (nothing evaluates to True)
```

#### 9. enumerate() - Adds a counter/index to your loop

```python
languages = ("Python", "Java", "C++")
print("\n--- Enumerating over tuple ---")
# Enumerate returns pairs of (index, element) which we unpack in the loop
for index, lang in enumerate(languages, start=1):
    print(f"Rank {index}: {lang}")
```

### 8. **Methods of Tuples*
#### Because tuples are` immutable`, they lack methods like *.append(), .pop(), or .sort()*. In fact, they only have **two** built-in methods:

### 1: `.count(value)` 
#### Counts how many times a `specific value` appears inside the tuple.


```python
t = (1, 2, 3, 2, 4, 2)
print(t.count(2))  # Output: 3
```

### `.index(value)`
#### Searches for a value and `returns the index position` of its `first occurrence`. Throws a **ValueError** if the item `isn't there`.


```python
t = ('a', 'b', 'c', 'b')
print(t.index('b'))  # Output: 1
```
