# A boolean is a `value` that is either **True** or **False** nothing else. 
## It's how a program answers `yes/no` questions: 
### "Is the user logged in?", "Is the score above 40?" 
#### Booleans are the foundation of every if statement you will ever write.

```python
is_login = True
is_logout  = False
print(type(is_login))   # <class 'bool'>
print(type(is_logout))  # <class 'bool'>
```

# None is different it means **"no value at all"**, an empty box. 
## It is not True, not False, not 0, not an empty string. 
### Use it when something hasn't been set yet.

```python
graduate = None # no decided yet
print(graduate)         # None
```

# note:
### Capitalization matters. Python needs `True` , `False` , and `None` with a `first capital letter`.
#### Writing true or none is an error.
