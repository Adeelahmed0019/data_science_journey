# Truthy & Falsy Values
## Here's a Python specialty: every value can be treated as `True` or `False` when used in a condition, even if it isn't a real boolean. 
### Python asks "does this feel empty or filled?"
### Falsy (act like False): the number 0 , an empty string "" , an empty list [] , an empty dict {} , and None.
### Truthy (act like True): basically everything else — any non-zero number, any non-empty string or list

```python
name = ""
if name:
    print("Hello")
else:
    print("Name is empty")   # this runs — "" is falsy


items = [1, 2]
if items:
    print("Cart has items")   # runs — non-empty list is truth
```
