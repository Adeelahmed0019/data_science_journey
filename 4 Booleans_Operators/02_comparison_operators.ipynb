# **Comparison Operators**
### Comparison operators ask a question and always answer with a `boolean True or False` 

### 1: == equal to 5 == 5 True
### 2: != not equal to 5 != 3 True
### 3: >greater than 7 > 10 False
### 4: < less than 7 < 10 True
### 5: >= greater than or equal 10 >= 10 True
### 6: <= less than or equal 9 <= 8 False

```python
marks = 75

# Equal to
print (marks == 75)     # True

#not equal to
print(marks != 75)      # False

# Greater than
print(marks > 50)       # True

# less than
print(marks < 100)     # True

# Greater than or equal to
print(marks >= 75)     # True

# less than or equal to
print(marks <= 100)    # True
```

# The most **Important** `interview` quastion
## How is a string compared with another string just by using the "==" equal to operator?

# **`Answer`**
## Python compares the two strings `character by character`, from `left to right`, based on their underlying numerical values `(ASCII/Unicode values)`.

### For two strings to be equal (True), they must satisfy three strict conditions.
#### 1: Exact Match: Every single character in the exact same position must be identical.
#### 2: Case Sensitivity: Capital letters and lowercase letters have different numerical values, so "A" == "a" is False. 
#### Same Length: The strings must have the exact same number of characters, including spaces.
