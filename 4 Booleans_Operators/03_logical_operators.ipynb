# Logical Operators: `and`, `or`, `not`
## These combine boolean questions into bigger ones.

### 1: and → `True` only if `both sides` are `True`.
### 2:  or → `True` if `at least one side` is `True`.
### 3: not → flips a boolean: not True is False

```python
age = 18
eligible = True

# For 'AND' both must be true to enter
print(age >= 18 and eligible)   # True

# For 'OR', any one true is enough
print(age < 13 or eligible)      # True

# For 'NOT', it reverses the boolean value
print(not eligible)               # False 
```
