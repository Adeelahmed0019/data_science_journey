##### 1: Create two `numbers` and print the result of all six `comparison operators`(==,!=, >, <, >=, <=) between them.

```python
num_1 = int(input("Enter first number: "))
num_2 = int(input("Enter second number: "))
print("num_1 is ==> ",num_1,"and num_2 is ==> ",num_2)
print("num_1 is equal to num_2: ",num_1 == num_2)
print("num_1 is not equal to num_2: ",num_1 != num_2)
print("num_1 is greater than num_2: ",num_1 > num_2)
print("num_1 is less than num_2: ",num_1 < num_2)
print("num_1 is greater than or equal to num_2: ",num_1 >= num_2)
print("num_1 is less than or equal to num_2: ",num_1 <= num_2)
```

##### 2: is_raining = True and has_umbrella = False. Print whether the person will stay dry (they stay dry if it's not raining OR they have an umbrella)

```python
is_raining = True
has_umbrella = False

if is_raining and not has_umbrella:    # Here "not" is used to reverse the boolean value of "has_umbrella"
    print("You are protected from the rain!")
else:
    print("You might get wet!")
```

### 3: Given a variable that could be a `name` or `an empty string`, use short-circuit or to print the name, or "Guest" if it's empty: print(name or "Guest") . Explain in a comment why this works.
Hint: or returns the first truthy value

```python
name = input("Enter name")  # This could be "any string" or an empty string " "

print(name or "Guest") # works because Python's 'or' operator evaluates from left to right and returns,
                      # the first 'truthy' value it finds. If 'name' is empty, it is considered 'falsy', 
# so Python 'short-circuits' and returns the second value, "Guest".
print(name or "Guest")
```

#### Hint: == compares values, is compares identity (same object in memory).

```python
a = 10
b = 20
c = a
print(a == b) # if the values of `a` and `b` are equal.

print(a is b) # if `a` and `b` refer to the `same object in memory`.

print(a is c) # if `a` and `c` refer to the same object in memory. 
```
