# Short-Circuit Evaluation

## Python is lazy in a smart way. With `and` and `or` , it stops checking as soon as it knows the answer.

### And stops at the `first False`, if the `left side` is *False*, the whole thing is `already False`, so Python never `looks at` the `right side`. 
### Or `stops` at the `first True`, if the `left side` is *True*, the answer is `already True`.

```python
def check():                # function to check if the right side is evaluated
    print("check ran!")     # print a message to indicate that the function was called
    return True             # return True to indicate that the right side is True

# left side is False, so check() is NEVER called
print(False and check())   # False  (no "check ran!")

# left side is True, so check() is NEVER called
print(True or check())    # True   (no "check ran!")
```
