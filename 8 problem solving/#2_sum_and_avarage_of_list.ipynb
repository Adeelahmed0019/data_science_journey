## `sum` and `average` of a **list** 
### Take marks and `Fine` the **sum* and ***Average**

```python
marks = [85, 90, 78, 92, 88]
sum = 0
for mark in marks:
    sum = sum + mark
average = sum / len(marks)
print("The total marks is:", sum)
print("The average marks is:", average)
```

## From Google

```python
# 1. Take a space-separated string as input from the user
user_input = input("Enter a list of numbers separated by spaces: ")

# 2. Convert the string into a list of strings
string_list = user_input.split()

# 3. Convert the list of strings into a list of integers (or floats)
numbers = [int(num) for num in string_list]

# Check if the user entered any numbers to avoid division by zero
if len(numbers) > 0:
    # 4. Manually calculate the sum using a loop instead of sum()
    total_sum = 0
    for num in numbers:
        total_sum += num

    # 5. Calculate the average
    average = total_sum / len(numbers)

    # 6. Display the results
    print(f"\nYour list: {numbers}")
    print(f"Sum: {total_sum}")
    print(f"Average: {average}")
else:
    print("You did not enter any numbers.")
```
