# 1st problem 
## print the numbers from `1` to `50` but 
### multiple of **5** ==> print **`Fizz`** 
### multiple of **3** ==> print **`Buzz`** 
### multiple of **3** and **5** ==> print **`FizzBuzz`** 
### otherwise ==> print **`the number`** 

```python
user_input = int(input("Enter any number chech it a number or multiple of '3' or '5'"))
for i in range(user_input+1):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```
