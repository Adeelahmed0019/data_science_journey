# 1 **Floating Point Precision**

```python
print(0.1 + 0.2 == 0.3)
```

# Why does this happen?
### Computers store numbers in binary (base 2). Just like 1/3 in decimal is 0.33333... (never ends), 0.1 in binary is 0.0001100110011... (also never ends). Since the computer has only 64 bits to store a float it rounds off, creating a tiny error.

## What's actually stored:

### Here is how the syntax breaks down:
#### 0.1: The actual number you want to format.
#### : Separates the value from the formatting rules.
#### .20: Specifies the precision (how many digits to show after the decimal point)
#### .f: Specifies the type, which stands for Fixed-point notation (displays the number as a standard decimal rather than scientific notation).


```python
a = 0.1 
b = 0.2
c = a + b
print("The value of 'a' is => ",a)
print("The value of 'b' is => ",b)
print("The value of 'C' is: => ",c)
print(f'The value of "C" is: => {c:.3f}')  # 0.30000000000000004441

# See the REAL stored values (10/20/30 decimal places)
print(f"{0.1:.10f}")  # 0.10000000000000000555
print(f"{0.2:.20f}")  # 0.20000000000000001110
print(f"{0.3:.30f}")  # 0.29999999999999998890

print(f"{0.1:.10f} + {0.2:.20f} = {c:.30f}") 
```

# 2

```python
# '/' always returns float — even when it divides evenly!
print(10 / 2)      # 5.0  (not 5!)

# '//' floors DOWN — watch with negatives!
print(-7 // 2)     # -4  (not -3!) cause it rounds down to the next lower integer
 
# '%' gives remainder — useful for even/odd, last digit
print(17 % 2)      # 1 (odd)

print(9876 % 10)   # 6 (last digit!)
```

### 3 **`divmod()` — Two answers in one shot**

```python
# 135 minutes = ? hours and ? minutes
hours, mins = divmod(135, 60)
print(hours, mins)  # 2, 15
```

# 5 Surprise 1: round(2.5) = 2, not 3!
#### **round()** it round a decimal number as `even` from .0 to .5 and from .6 to .9 round it as `odd` 


```python
print(round(2.5))  # 2  (rounds to nearest EVEN)
print(round(3.5))  # 4  (rounds to nearest EVEN)
print(round(4.4))  # 5  (rounds to nearest EVEN)
print(round(5.0))  # 5  (rounds to nearest EVEN)
```

# 6 Surprise 2: Day of the week using %


```python
# Today is Monday. What day is it after 100 days?
number_of_days = int(input("Enter the number of days ahead: "))
remainder = number_of_days % 7
if remainder == 0:
    print("The day is: Monday")
elif remainder == 1:
    print("The day is: Tuesday")
elif remainder == 2:
    print("The day is: Wednesday")
elif remainder == 3:
    print("The day is: Thursday")
elif remainder == 4:
    print("The day is:  Friday")
elif remainder == 5:
    print("The day is: Saturday")
elif remainder == 6:
    print("The day is: Sunday")
else:
    print("Error: remainder is not in the range 0-6")
```
