# **String slicing**
### `Slice` means **part of something** so the slicing is the process of `Extracting part of the` **String**
#### Remember that: in slicing `indexing` matters a lot
#### Forward indexing = 0, Backward indexing = -1
#### `Syntex of slicing`: srt[start : stop : step ]


##### `Remember that while slicing always keep indexing in your` **MIND** `because in slicing we do not follow the sequence of the string we follow the` **indexing**

## Simple indexing

```python
str = "wellcome"
print("The length of the string is: ",len(str)) # this will print the length of the string
for i in range(len(str)): # this will print the index and the character at that index in the string
    print(i,"-->", str[i])  # this will print the index and the character at that index in the string
   
# Explaination of the code:
# 1: We have a string "wellcome"
# 2: "for" loop is used to iterate over the range of the length of the string
# 3: "i" is the index of the character in the string
# 4: 'in' keyword is used to iterate over the range of the length of the string
# 5: "len(str)" is used to get the length of the string
# 6: 'print()' function is used to print the index and the character at that index in the string
# 7: inside the print() function, 'i' is the index and 'str[i]' is the character at that index in the string
```

## `enumerate` for set *indexing* manually

```python
str = "wellcome"
for index , char in enumerate(str, start=1): # this will print the index and the character at that index in the string
    print(index, ":",char)    
```

# **Slicing**

### Positive `indexing`

```python
# With Starting indexing 0 

str = "wonderful"
print("This is our string ==>" , str)

for i in range(len(str)):
    print("indexing: ",i,":", str[i]) 
    
# Explaination of the code:
# 1: We have a string "wonderful"
# 2: "for" loop is used to iterate over the range of the length of the string
# 3: "i" is the index of the character in the string
# 4: 'in' keyword is used to iterate over the range of the length of the string
# 5: "len(str)" is used to get the length of the string
# 6: 'print()' function is used to print the index and the character at that index in the string
# 7: inside the print() function, 'i' is the index and 'str[i]' is the character at that index in the string
```

### Negative `indexing`

```python
print("This string is:  ==>", str)
str = "wonderful"

j = len(str)                                
print("The length of the string is: ",j) 
for i in range(j):                          
    print("indexing: ",i,":", str[i], j)   
    j-=1                                 
    
# Explaination of the code:
# 1: We have a string str = "wonderful"
#2: "j" is used to hold the length of the string which is 9 in this case
# 3: "for" loop is used to iterate over the range of the length(9) of the string = 9
# 4: "i" is the index(0 to str - 1) of the character in the string
# 5: 'in' keyword is used to iterate over the range of the length of the string
# 6: range(j) is used to iterate over the range of the length(9) of the string
# 7: 'print()' function is used to print the index and the character at that index in the string
# 8: inside the print() function, 'i' is the index and 'str[i]' is the character at that index in the string
# 9: j -= 1 it will decrement the value of j(9) by 1 in each iteration, 9-1 = 8, 8-1 = 7, 7-1 = 6, 6-1 = 5, 5-1 = 4, 4-1 = 3, 3-1 = 2, 2-1 = 1, 1-1 = 0
```

#### reverse the `string`

```python
str = "wonderful"
print("This is our string ==>", str)

# Here 'for' is just to print the index and the character at that index in the string
# NOTE: without 'for' loop we can directly slice the string using str[start:end] or str[start:end:step]

j = len(str)
for i in range(j):
    print("indexing: ",i,":", str[i], -j)
    j-=1

# Reverse the string using slicing
print(str[::-1])
# str[::-1]  # here we are slicing the string from the end to the beginning, 
# but since we are using negative indexing, it will slice until the first character of the string.
print(str[-1::-1])
```

```python
s = "PYTHON PROGRAMMING"
for i in range(len(s)):
    print(f"{i}:{s[i]}", end=" ")
s[::-2]
```

#### `2: [2:3]`

```python
s = "PYTHON PROGRAMMING"
print(len(s))
for i in range(len(s)):
    print(f"{i}:{s[i]}", end="  ")
s[2:3] # This will print the character at index 2, which is "T".
```

#### `3: [2:3:-1]`


```python
s = "PYTHON PROGRAMMING"
for i in range(len(s)):
    print(f"{i}:{s[i]}", end="  ")
s[2:3:-1] # because the step is -1, it will try to slice the string in reverse order, but since the start index (2) is less than the stop index (3), it will return an empty string.
```

#### `4: [-6:-16]`

```python
s = "PYTHON PROGRAMMING"
print(len(s))
for i in range(len(s)):
    print(f"{i}:{s[i]}", end="  ")
s[-6:-16:] 
# s[-6:-16] 
# because the step is not specified, it defaults to 1, which means it will slice the string from left to right.
# However, since the start index (-6) is greater than the stop index (-16), it will return an empty string.
# NOTE: by default, the step is 1, which means it will slice the string from left to right. 
# If we want to slice the string in reverse order, we need to specify the step as -1.
```
