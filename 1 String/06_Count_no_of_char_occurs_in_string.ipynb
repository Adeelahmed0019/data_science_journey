# *Count the Number of Character Occurs in the String*

```python
# 1st step: User input for a string
String = input("Enter a string: ".lower())

# 2nd step: Display the entered string
print("The string you entered is:", String)

# 3rd step: User input for the characters to check frequency [N_of_C = number of characters]
N_of_C = input("Enter the characters to check frequency : ")

# 4th step: Loop through check the frequency of character in the string
count = 0
for char in String:
    # If the character matches the input character, increment the count
    if char == N_of_C:
        count += 1
# 5th Step  Display the frequency of the character in the string
print(f"The frequency of '{N_of_C}' in the string is: {count} times.")
```
