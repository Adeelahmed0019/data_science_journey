# Reverse the `Alternative` words from the given list

```python
# 1st Step: Take input from the user in string format
string = input("Enter a string: ")

# 2nd Step: Split the string into words and store it in a list
words = string.split() # in list format

# 3rd Step: Iterate on list to extract alternate words starting from index 1
for index in range(1, len(words) ,2): 

    # 4th Step: Initialize a list to store the reverse of the alternate word
    rev = ''                         
 
 # 5th step: Iterate on the characters of the alternate word to reverse it and store it in reverse variable 
    for ch in words[index]:           
        
        # 6th Step: Concatenate the characters in reverse order to form the reversed word and store it in reverse variable 
        rev = ch+rev                  
 
 # 7th Step: Replace the alternate word with its reverse in the list
    words[index] = rev                
    
 # 8th Step: Join the list of words to form a string and display the result   
reverse = ' '.join(words)

# 9th Step: Display the string after reversing the alternate words
print("After revesing the string of alternate words : --> ", reverse)
```
