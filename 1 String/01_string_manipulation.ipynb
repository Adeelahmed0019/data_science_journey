# **String Manipulation**

### **Definition**
### String is the sequence of character `Enclosed` in `Single`, quotes `Double` quotes and `Triple` quotes
### String are the `Immutable` they can not be `Changed`
### In string each character remains at a `Unique` position, or index
### Index goes from `0` to length `-1`(in forward direction) and `-length` upto `-1` in backward direction


# **Traversal of a string**
### Traversing means: to accessing the `Individual` character of the string(from first to last)
#### "For loop" can be used to `traverse` the string

# **Indexing**
### Indexing means: every character has a different `position`(in back-hand) from `0` to `size -1` 

# `Strings` in different `Quotes`

### In single quotes

```python
# in single quotes
S1 = 'string printed in single quotes'
print('string printed in single quotes')
print(type(S1))
print("Total length of the 'string' is: ",len(S1))
```

### In double quotes

```python
# in double quotes
S2 = "string printed in double quotes"
print("string printed in double quotes")
print(type(S2))
print("Total length of the 'string' is: ",len(S2))
```

### In triple quote with single quotes

```python
# in triple quote with single quotes
S3 = '''string printed in triple quote with single quotes'''
print('''string printed in triple quote with single quotes''')
print(type(S3))
print("Total length of the 'string' is: ",len(S3))
```

### In triple quote with double quote

```python
# in triple quote with double quote
S4 = """string printed in triple quote with double quotes"""
print("""string printed in triple quote with double quotes""")
print(type(S4))
print("Total length of the 'string' is: ",len(S4))
```

# String `indexing`

```python
# string
S1 = 'string printed in single quotes'

# forward:
# indexing starts from 0, so S1[1] will give the second character of the string
print(S1[1]) 
 
# backward
# negative indexing starts from the end of the string
print(S1[-1]) 

# checking `ID` of the string, id means the memory address where the variable is stored
s = 'aaa'
print(s, id(S1)) 
s = 'bbb'
print(s, id(S1))

```

## String **Traversing** by 'For loop'

```python
# simple Traversing
string  ="wonderful"

# here `i` is a variable it will access string's variable element one by one
for i in string: 
    print(i, end=' ') # end=' ' is used to print the characters in the same line with one space between the characters
print("\nThe length of the this 'string' is:",len(string))
```

## Traversing with `Range` function

```python
# Traversing with `Range` function

string_rng = "python"
# here `i` is a variable it will access string's variables index one by one and range function will give the index values from 0 to length of the string -1
for i in range(0,len(string_rng)): # range takes two values fist starting[0 index], and second end[-1 index] [5-1= 4] 
    print(string_rng[i], end="")
print("\n")



```

## Traversing with `enumerate` function 
Instead of returning just the item itself, enumerate() wraps any iterable (like a list, string, or tuple) and returns it as a series of tuples. Each tuple contains two elements: (index, item)

```python
# srting:
srt = "py"

print(list(enumerate(srt)))# list/tuple of index and character of the string


# enumerate function will give the index and the character of the string one by one
string_rng = "python"
for index, char in enumerate(string_rng):
    print("Index", index, "is", char)
    
# Explaination of the above code:

#1: "for" is a loop that will iterate over the string "python" one character at a time. 
#2: "index" and "char" are two variables that will hold the index and character of the string respectively.
#3: "enumerate(string_rng)" is a built-in function that will return the index and character of the string one by one.
#4: "print("Index", index, "is", char)" will print the index and character of the string one by one.
```

## `enumerate` practical example: 
 1. Categorical Encoding (Label Encoding)

```python
categories = ['Low', 'Medium', 'High', 'Critical', 'Very High', 'Very Low']
encoding_map = {category: i for i, category in enumerate(categories)}
print(encoding_map)


# Code Explanation:

#1: 'categories': is a "list"(variable) of categories that we want to encode(categorical to numerical)
#2: 'encoding_map': is a 'dictionary' that maps each category to its corresponding index in the list
#3: inside the dictionary: The first 'category' is the key and fisrt 'i' is the value of the category in the list.
#4: Here first 'i' is the index of the category in the list and second 'i' is the value of the category in the list.
#5: The 'enumerate' function is used to get both the index and the value of each category in the list.
```

# Print String in `Reverse` order

```python
string_reverse = "Python"
print(string_reverse)

for i in range(-1 , -len(string_reverse) -1, -1):
    print(string_reverse[i], end='')

# Explanation of negative indexing:

# Length of string is  = 6
# Index of string is  = 0 to 5
# Negative index of string is = -1 to -6 [-6 , -5 , -4, -3 , -2, -1]
# Here we are starting from -1/n
# And ending at -len(string_reverse)-6/P 
# with step -1(one) [n to p step by step]

# Explanation of range(-1 , -len(string_reverse) -1, -1):

# first value -1 means start from last index of the string which is -1/n
# second value -len(string_reverse) means end at the first index of the string which is -len(string_reverse)/p
# third value -1 means move one step in reverse direction
```

```python
# Explanation of negative indexing and range function for reverse traversing
a = "Python"
print(len(a)) # 6
print(-len(a))  # -6
print(a[-1]) # last 'n' character will print first because -1 is the index of 'n' character in the string.
print(a[-6]) # first 'P' character will print first because -6 is the index of 'P' character in the string.
```
