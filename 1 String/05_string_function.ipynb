### **isalnum()** [is alpha-numeric]
##### It will return `True` if the string contains only `letters` or `digits`
##### It returns `False` if any `special` character occurs even ``

```python
srt1 = "abc123"
str2 = "abc 123"
print(srt1.isalnum()) # it return `True` bc there is no space in srt1
print(str2.isalnum()) # it return `False` bc there is space in str2
```

### **isalpha()** 
##### It retuns `True` if all the character in the `string` are `alphabets` otherwise return `False`

```python
srt_isalpha = "abc"
srt_isalpha_2 = "asd@" 
print(srt_isalpha.isalpha()) # it return `True` bc there is only alphabet in srt_isalpha
print(srt_isalpha_2.isalpha()) # it return `False` bc there is `@` in srt_isalpha_2
```

### **isdigit()**
##### `inverse` of the `isalpha()`

```python
srt_digit = "123"
srt_digit_2 = "123a"
print(srt_digit.isdigit()) # it return `True` bc there is only digit in srt_digit
print(srt_digit_2.isdigit()) # it return `False` bc there is `a` in srt_digit_2
```

### **lstrip()**
##### Returns the `String` after `removing` the `space(s)` on the `left` of the string 
##### And also remove the character from left side that you passed in *Istrip()* function
####

```python
# Remove space from left side of the string

srt_lstrip = "  data science   "

# This is orignal string 
print("orignal string 'str_1strip is ' ==>",srt_lstrip)

# The length of string is 17 
print("\nThe length of srt1 is : ",len(srt_lstrip))

# it remove space from left side of the string
print("After applying 'lstrip function' on ste_1strip: ", srt_lstrip.lstrip())

# The length of string is 15 because there are two space in left side of the string 
print("The length of string after applying 'lstrip function' on str_1strip is : ",len(srt_lstrip.lstrip()))

print('*'*100)

#########################################################################################################################

# Remove character 'd' from left side of the string


srt_lstrip2 = "dddata science"

# This is orignal string
print("orignal string 'str_2strip is ' ==>", srt_lstrip2)

# The length of string is 13
print("\nThe length of srt2 is : ",len(srt_lstrip2))

# it remove `d` from left side of the string
print("After applying 'lstrip function' on str_2strip: ", srt_lstrip2.lstrip("d"))

# The length of string is 12 because there are one `d` in left side of the string
print("The length of string after applying 'lstrip function' on str_2strip is : ",len(srt_lstrip2.lstrip("d")))
```

### **rstrip()** 
##### Returns the `String` after `removing` the `space(s)` on the `right` of the string 
##### And also remove the character from right side that you passed in *rstrip()* function
####

```python
# Remove space from left side of the string

srt_rstrip = "data science   "

# This is original string 
print("original string 'str_1strip is ' ==>",srt_rstrip)

# The length of string is 17 
print("\nThe length of srt1 is : ",len(srt_rstrip))

# it remove space from right side of the string
print("After applying 'rstrip function' on srt_1strip: ", srt_rstrip.rstrip())

# The length of string is 15 because there are two space in right side of the string 
print("The length of string after applying 'rstrip function' on str_1strip is : ",len(srt_rstrip.rstrip()))

print('*'*100)

#########################################################################################################################

# Remove character 'd' from left side of the string


srt_rstrip2 = "data sciencewww"

# This is original string
print("original string 'str_2strip is ' ==>", srt_rstrip2)

# The length of string is 13
print("\nThe length of srt2 is : ",len(srt_rstrip2))

# it remove `w` from right side of the string
print("After applying 'rstrip function' on str_2strip: ", srt_rstrip2.rstrip("w"))

# The length of string is 12 because there are one `w` in right side of the string
print("The length of string after applying 'rstrip function' on str_2strip is : ",len(srt_rstrip2.rstrip("w")))
```

### **isspace()**
#### This fuction returns `true` only if a string `only` contains `spaces`
#### This function returns `false` only when a string is containing any value `except` *spaces*

```python
isspace_srt = " "
not_isspace_srt = " data science  "

# It return `True` bc there is only space in isspace_srt
print("original string is ==>", isspace_srt, ":: output is: ", isspace_srt.isspace()) 

# It return `False` bc there is not only space in not_isspace_srt
print("original string is ==>", not_isspace_srt, ":: output is: ", not_isspace_srt.isspace()) 
```

### **join()**
##### Join function `joins` the list items with separator

```python
simple_str = ["Jan", "Feb", "Mar"]

print("original string is : ", simple_str)
print("The data type of 'simple_str is : ' ",type(simple_str))

separator = ", "
print("The data type of 'separator is : ' ",type(separator))

new = separator.join(simple_str)
print("After 'join' funtion : ", separator.join(simple_str))
print("The data type of 'new' is : ' ",type(new))
```

#### **swapcase()**
##### This function `swap` the string lower-case to uper-case and wise-varsa

```python
swap_str = "ADEEL"
swap_str2 = "adeel"
print("The original string is :: ", swap_str)
print('After applying "swapcase()" function :: ', swap_str.swapcase())
print("The original string is :: ", swap_str2)
print('After applying "swapcase()" function :: ', swap_str2.swapcase())
```
