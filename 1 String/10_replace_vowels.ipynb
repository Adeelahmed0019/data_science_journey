# Replace Vowels with ==> '*'

```python
# user input a string and replace all the vowels with asterisks (*)
string = input("Enter a string: ")

# replace all the vowels with asterisks (*) and print the new string
new_string = ""

# loop through each character in the string and check if it is a vowel or not.
for i in string:
    # if it is a vowel, replace it with an asterisk (*), otherwise keep it as it is.
    if i in "aeiouAEIOU":
        new_string += "*"
# print the new string with all the vowels replaced by asterisks (*).
    else:
    # if it is not a vowel, keep it as it is.
        new_string += i
print(new_string)
```
