# Capitalized `First` and `last` letter of each `word` in a given string

```python
# 1st Step: Take input from the user
string = input("Enter a string:")

# Display the original string
print("The original string is ==> ",string)
print("The data type of the string is:", type(string))
print("*" * 50)
# 2nd Step: Make first letter of each word uppercase and the rest lowercase

# Title is used to convert the first letter of each word to uppercase and the rest to lowercase
string = string.title()  
print("The string with first letter of each word in uppercase is ==> ", string)
print("The data type of the string is:", type(string))
print("*" * 50)
# 3rd step: Split the string into a list of words using the split() method
splitted_string = string.split()  
print("The splitted string is ==> ", splitted_string)
print("The data type of the splitted string is:", type(splitted_string))
print("*" * 50)
empty_string = "" # 4th step: Loop through the list of words. NOTE: splitted_string is a list of words. Each word is a string.
for word in splitted_string:
    empty_string = empty_string+word[:-1]+word[-1].upper()+" "
print("First and last char of each char is upper_case:", empty_string)

# explaining the code:
# 1. The code takes a string input from the user.
# 2. It converts the first letter of each word in the string to uppercase and the rest to lowercase using the title() method.
# 3. It splits the string into a list of words using the split() method.
# 4. loop working
#   - It initializes an empty string called empty_string.
#   - It loops through each word in the splitted_string list.
#   - word[:-1] takes all characters of the word except the last one.
#   - word[-1].upper() takes the last character of the word and converts it to uppercase.
#   - It concatenates the modified word to the empty_string with a space in between.
```
