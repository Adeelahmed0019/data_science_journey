# Check the string is `Palindrome` or not

## 1st Method (simplest)

```python
# Take user input as string.
Original = input("Enter a string: ".lower())

# reverse the string using slicing.
reversed_string = Original[::-1]

# Check if the 'original' string is equal to the "reversed" string.
if Original == reversed_string:
    print(f'"{Original}" is a palindrome.')
else:
    print(f'"{Original}" is not a palindrome.')
```

## 2nd Method (For loop)

```python
# Take input from the user
Original = input("Enter a string: ".lower())

# Initialize an empty string to store the reversed version of the original string
reversed = ''

# Use a for loop to iterate through the original string in reverse order and create the reversed string
for i in range(len(Original) -1, -1, -1):
    reversed = reversed+Original[i]

# Check if the original string is equal to the reversed string
if Original == reversed:
    print(Original, 'is a palindrome.')
else:
    print(Original, 'is not a palindrome.')
```

## 3rd Method (Function)

```python
 # Define a function to check if a string is a palindrome or not
def is_palindrome(s):
    
    # Remove spaces and convert to lowercase
    s = s.replace(" ", "").lower() 
    
    # Check if the string is equal to its reverse
    return s == s[::-1]
# Get user input
user_input = input("Enter a string: ")

# Check if the input string is a palindrome
if is_palindrome(user_input):
    print(f"'{user_input}' is a palindrome.")
else:
    print(f'"{user_input}" is not a palindrome.')
```
