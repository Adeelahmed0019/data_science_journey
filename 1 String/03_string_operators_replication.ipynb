# **String operators and replication**

### Basically there are **four** string `opertors`
#### 1 **+ or concatenation** (Joining and combining)
#### 2 **`*` multipliction** (Repetition of string)
#### 3 **membership** `in` and `not in`
#### 4 **Comparison** operator

# 1 **String + concatenation**
### Note: You can't `add` number and string

```python
a = 'tea'                   # this is a string
b = '121'                   # this is also a string

c = a + b                   # this is also a string because we are concatenating two strings
print(c)                    # this will print 'tea121'

print(type(c))              # this will print <class 'str'> because c is a string

a = 'tea'                   # this is a string
b = '121'                   # this is also a string

print(a + b)                # this will also print 'tea121' because we are concatenating a and b 

print('1' + '2')            # this will print '12' because we are concatenating two strings '1' and '2'
print("abc" + "123")        # this will print 'abc123' because we are concatenating two strings 'abc' and '123'
 
```

# 2 **String * Replication**
## Note: you can't multiply `string` and `srting` using `*` only `num * str` or `num*num` 

```python
print('Hi!'* 2)         # this will print 'Hi!Hi!' because we are repeating the string 'Hi!' 2 times
print(2* "Hellow")      # this will print 'HellowHellow' because we are repeating the string 'Hellow' 2 times
print('2' * 5)          # this will print '22222' because we are repeating the string '2' 5 times
```

# 3 `Membership Operators`
#### The `membershp` operator  `in` and `not in` are used to check the  `presence` of the character in the string 

```python
print('a' in 'python')         # return False because 'a' is not present in the string 'python'

print('a' in 'java')           # return True because 'a' is present in the string 'java'

print('per' in 'operators')    # return True because 'per' is present in the string 'operators'

print('Man' in 'manipulation') # return False because 'Man' is not present in the string 'manipulation' (case sensitive)

print('pre' not in 'presence') # this will return False because 'pre' is present in the string 'presence'

print("ee" in "adeel") # this will return True
```

# 4 `Comparison operator`
#### We can apply comparison operator (== , != , > , < , >= , <= )
#### The comparison will be **character** by **character**
#### `NOTE: if we compare two same string we are comparing just` **ASCII code**

```python
# Comparison operators with strings

# Equal to (==)
print('abc' == 'abc')         # return True because both strings are the same

# Not equal to (!=) 
print('abc' != 'def')         # return True because 'abc' is not equal to 'def' 

# Greater than (>)
print("abc" > "abc")   # return False because both strings are the same
print("abd" > "abc")   # return True because 'abd' is greater than 'abc' (lexicographical order)

# Less than (<)
print("abc" < "abd")   # return True because 'abc' is less than 'abd'

# Greater than or equal to (>=)
print("abd" >= "abc")   # return True because 'abd' is greater than 'abc' (lexicographical order)
 
# Less than or equal to (<=)
print("abc" <= "abd")   # return True because 'abc' is less than 'abd' (lexicographical order)
```

## Check the `ASCII` code values

```python
# ASCII values character[A to Z] to integer  A to Z is 65 to 90 
# ASCII values character[a to z] to integer  a to z is 97 to 122
# ASCII values character[0 to 9] to integer  0 to 9 is 48 to 57
```

```python
# ASCII values character to integer, chr() function returns the character of an ASCII value
print("65: ",chr(65))    # this will return 'A' because 65 is the ASCII value of 'A'
print("90: ",chr(90))    # this will return 'Z' because 90 is the ASCII value of 'Z'
print("97: ",chr(97))   # this will return 'a' because 97 is the ASCII value of 'a'
print("122:",chr(122))   # this will return 'z' because 122 is the ASCII value of 'z'
print("48: ",chr(48))   # this will return '0' because 48 is the ASCII value of '0'
print("57: ",chr(57))   # this will return '9' because 57 is the ASCII value of '9'
```

```python
# ASCII values from characters, ord() function returns the ASCII value of a character
print("A: ",ord('A'))    # this will return 65 because 'A' is the ASCII character for 65
print("Z: ",ord('Z'))    # this will return 90 because 'Z' is the ASCII character for 90
print("a: ",ord('a'))    # this will return 97 because 'a' is the ASCII character for 97
print("z: ",ord('z'))    # this will return 122 because 'z' is the ASCII character for 122
print("0: ",ord('0'))    # this will return 48 because '0' is the ASCII character for 48
print("9: ",ord('9'))    # this will return 57 because '9
```

```python
# ASCII values of digits 0 to 9, ord() function returns the ASCII value of a character
print(ord("0"))   # this will return 48 because '0' is the ASCII character for 48
print(ord("9"))   # this will return 57 because '9' is the ASCII character for 57
```
