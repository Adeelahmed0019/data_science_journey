# **1: LIST**

### A list is an ordered, changeable (mutable) collection of data items that can hold items of any type
# **Features of list**
### List holds `multiple-values` of `multiple data types` 
### List is an Ordered.
### Mutable
### It allows `duplicate` values

### 1. **Creating Lists**
#### create the list
#### check the `data types` of the list
#### check the `length` of the list
#### print the list

```python
# Empty list
items = []                  # square brackets denote the list(this is an empty list)

# check the data type of items.
print(type(items))          # <class 'list'>, type() function returns the type of the object, here it will return list

# check the length of items.
print(len(items))           # len() function returns the number of items in a list, here it will return 0

# print the items in the list.
print(items)                

# List of numbers
marks = [85, 90, 72, 60]    # list having 4 integer values

# check the data type of marks.
print(type(marks))            # <class 'list'>

# check the length of marks.
print(len(marks))            # len() function returns the number of items in a list, here it will return 4

# print the items in the list.
print(marks)                 


# Mixed data types allowed
mixed = [1, "Adeel", 3.14, True] # list having integer, string, float and boolean values

# check the data type of mixed.
print(type(mixed))            # <class 'list'>

# check the length of mixed.
print(len(mixed))             # len() function returns the number of items in a list, here it will return 4

# print the items in the list.
print(mixed)                  # prints the mixed list


# Using the list() constructor, constructor is a built-in function that creates a new list object
letters = list("python")   # ['p','y','t','h','o','n']

# check the data type of letters.
print(type(letters))          # <class 'list'>

# check the length of letters.
print(len(letters))           # len() function returns the number of items in a list, here it will return 6

# print the items in the list.
print(letters)                # prints the list of letters
```

## 2: Indexing          
### Each item has a position. Indexing starts at 0. Negative indexes count from the end.

```python
numbers = [10, 20, 30, 40, 50, 60]

for index, element in enumerate(numbers): # enumerate() function returns both the index and the element of the list
    print(f"Index {index}: Element {element}")
```

```python
# creating a list with different data types
my_list = [1, 2, 3, 4, 5, "Adeel", 3.14, True]  # list having integer, string, float and boolean values

print("The len of list is:",len(my_list))  # prints the length of the list

# accessing elements of the list using positive indexing
print(my_list[0])  # accessing the first element of the list
print(my_list[1])  # accessing the second element of the list

# accessing elements of the list using negative indexing
print(my_list[-1])  # accessing the last element of the list
print(my_list[-8])  # accessing the second last element of the list
```

## 3: **Slicing**
### Grab a portion of a list with  list[start : stop : step] . start is included, stop is excluded.


```python
nums = [10, 20, 30, 40, 50, 60] # list having 6 integer values
print(nums[1:4])    # prints the elements from index 1 to index 3 (4 is not included)
print(nums[:3])     # prints the elements from the beginning to index 2
print(nums[3:6])    # prints the elements from index 3 to index 5
print(nums[::2])    # prints every second element
print(nums[::-1])   # prints the list in reverse order
print(nums[1:5:2])  # prints every second element from index 1 to index 4
```

## 4: **Modifying Lists**

## 1: **Methods to Add Elements**

### 1:  **append(element)**: `Adds` a single element to the very `end of the list`. 


```python
# Initialize a list
fruits = ["apple", "banana"]

# print the original list
print("The original list is: ",fruits)

# Adding by using append()
fruits.append("cherry")      # ['apple', 'banana', 'cherry']

# print the list after adding an element using append()
print("List after appending element: ",fruits)
```

### 2: **extend(iterable)**: Adds `all elements` from another iterable (like another list) to the `end of the` current list.

```python
# Initialize a list
fruits = ["apple", "banana"]

print(f'The original list is: {fruits}')

# Adding by using extend()
fruits.extend(["date", "fig",22])# ['apple', 'banana', 'cherry', 'date', 'fig']

#print the list after adding an element using extend()
print(f'list after "Extend": {fruits}')
```

### 3: **insert()**: Puts an item `exactly where you want` it by shifting everything else to the right.

```python
# initialize a list
fruits = ['apple', 'cherry']

# The original list is:
print(f'The original list is: {fruits}')

# Adding by using `insert()`

fruits.insert(1, 'banana')  # Inserts at index 1
print(f"List after using `insert()`: {fruits}")  # Output: ['apple', 'banana', 'cherry']
```

## 2: **Methods to Removing Elements**

### 1: **remove()**: Deletes the `first matching` item it `finds by value`.

```python
# Initialize the list
numbers = [1, 2, 3, 3]
print(f'The original list is: {numbers}')

#  # After using 'remove()'
numbers.remove(2)  # Removes the first '2'
print(numbers)  # Output: [1, 3, 2]
```

### 2: **pop()**: `Cuts out` an item `by its index` and hands it `back to you`. If you `don't give` it an index, it takes the `last item`.

```python
# initialize the list
fruits = ['apple', 'banana', 'cherry','popcone']

print(f'The original list is: {fruits}')

# remove the item by using 'pop' without `indexing`
last_fruit = fruits.pop()  # Removes 'cherry'
print(f'List after using "pop()" without indexing: {last_fruit}')

# remove the item by using 'pop' with `indexing
first_fruit = fruits.pop(0)  # Removes 'apple'
print(f'List after using "pop(0)" without indexing: {first_fruit}')

print(f'The final list is: {fruits}')
```

### 3: **clear()**: `Wipes out` the `entire list` instantly.

```python
# Initialize the list
fruits = ['apple', 'banana']

# print the list:
print("The original list is: ", fruits)

# Using the 'clear()'
fruits.clear()
print("List after using 'clear()':",fruits) 
```

## 3: **Methods to Searching and counting Elements**

### 1: **index()**: Tells you the `position number of` the first matching item it finds

```python
# initialize the list
fruits = ['apple', 'banana','banana', 'cherry']
# print the list:
print("The list is:", fruits)
# After using index()
print(fruits.index('banana'))
```

### 2: **count()**: Tells you `exactly how many times` an `item shows up`.

```python
# Initialize the list:
letters = ['a', 'b', 'a', 'c', 'a']
# print the list:
print(f'The original list is: {letters}')

# after using `count()`
print(letters.count('a'))
```


 ## 4: **Methods to Ordering & Copying Elements**

### 1: **sort()**: `Organizes` your `list alphabetically` or `numerically` in place

```python
# initilize the list
numbers = [3, 1, 4, 2]

# print the list:
print(f'The original list is: {numbers}')

# After using `sort()` "ascending order"
numbers.sort()
print(f'After using "sort() as "Ascending order": {numbers}')

# After using `sort()` "decending order"
numbers.sort(reverse=True)  # Sorts descending
print(f'After using "sort() as "decending order": {numbers}')
```

### 2: **reverse()**: `Flips` the `current order` of the list `backwards`.

```python
# intialize the list:(words)
reverse_list = ['apple', 'banana', 'cherry']

# numeric
reverse_list = ['1,2,3,4,5']


# print the list:
print(f'The actual list is: {reverse_list}')

# After using `reverse()`
reverse_list.reverse()
print(f'list after "reverse()": {reverse_list}')  # Output: ['cherry', 'banana', 'apple']
```

### 3: **copy()**: Creates a `separate duplicate` of your list so `changes` to the new one `won't ruin the original`.

```python
# initilize the list
original = [1, 2, 3]
# print the list
print(f'The original list is: {original}')

# Copy the list
duplicate = original.copy()
#print the copied list
print(f'The copied list is: {duplicate}')

# Add new item in copied(duplicate) list
duplicate.append(4)

#print after adding new item in copied list
print(f'copied list after adding new item in {duplicate}')

# original
print(f'The actual list is: {original}')  

# copy and add
print(f'The list which is duplicate of "original" and add new item in it{duplicate}') 
```
