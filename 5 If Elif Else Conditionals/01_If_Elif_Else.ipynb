# 1: **What Are Conditionals & Why They Matter**
### Until now our programs ran top to bottom, every line, every time. Conditionals let a program choose what to do based on a situation, this is called `control flow`. It is the moment your code `stops` being a `calculator` and `starts` behaving like a `decision-maker`. 
## Real-life examples
##### • If marks >= 33  Pass, else Fail
##### • If balance is enough allow payment, else fi decline
##### • If user typed the right password  login, else show error
## Note: Python uses indentation (spaces) to know which lines belong inside an if. This is not optional in Python, wrong indentation = error. Use 4 spaces, stay consistent.

## 2: **if / elif / else Syntax**
### The core structure. if checks a condition; elif (else-if) checks more conditions; else catches everything left over.

```python
age = int(input("Enter your age: "))
if age < 13:
    print("You are a child")
elif age < 20:
    print("You are a teenager")
elif age < 60:
    print("You are an adult")
else:
    print("You are a senior citizen")
```

### How Python reads it: it checks conditions `top` to `bottom` and runs the `first` one `that is True`, then `skips` the `rest`. Order matters!

## The 4 rules to remember:
##### • Only one if per block, but as many elif as you want.
##### • else is optional, and takes no condition.
##### • The colon : at the end of each line is mandatory.
##### • Everything indented under a branch runs only if that branch is chosen.


## 3: **Nested Conditions**
#### An if inside another if. Use it when a second decision only makes sense after the first is True.

```python
logged_in = True                # anyone can login to the system, but only admin can access the admin panel
is_admin = False                # if user is admin, then they can access the admin panel, otherwise they cannot
if logged_in:                   # here we are checking if the user is logged in or not
    if is_admin:                # here we are checking if the user is an admin or not
        print("Welcome, Admin") # if the user is an admin, then they can access the admin panel
    else:
        print("Welcome, User")  # if the user is not an admin, then they cannot access the admin panel
else:
    print("Please log in first")
```

## 4: **One-Line if (Ternary Expression)**
#### A compact way to choose between two values in a single line. 
#### Pattern: value_if_true if condition else value_if_false

```python
# normal way
marks = 50

if marks >= 60:
    print("Pass")
    
else:
    print("Fail")

#one-line way (same thing)
result = "Pass" if marks >= 33 else "Fail"
print(result)
```

## 5: **match-case (Python 3.10+ Switch)**
#### When you compare ONE variable against MANY fixed values, match-case is cleaner than a long elif chain. (Other languages call this 'switch'.)

```python
day = input("Enter the day of the week: ")
days = day.lower()
match days:
    case "monday":
        print("First day of the week")
    case "saturday":
        print("First day of the weekend")
    case "sunday":
        print("last day of the weekend")
    case _:
        print("Invalid day of the week")
```

#### Note: Rule of thumb: a few conditions fi if/elif. Many fixed-value matches on one variable fi match-case.

## 6: **Real Project Grade Calculator**
#### Goal: take a student's marks (0–100) and print their grade.


```python
marks = int(input("Enter marks (0-100): "))
if marks < 0 or marks > 100:
    print("Invalid marks!")
elif marks >= 90:
    grade = "A+"
    print(f"Grade: {grade} - Outstanding")
elif marks >= 75:
    grade = "A"
    print(f"Grade: {grade} - Excellent")
elif marks >= 60:
    grade = "B"
    print(f"Grade: {grade} - Good")
elif marks >= 33:
    grade = "C"
    print(f"Grade: {grade} - Pass")
else:
    print("Grade: F - Fail")
```
