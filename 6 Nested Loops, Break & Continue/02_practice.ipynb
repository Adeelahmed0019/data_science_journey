## **Basic**
### 1: Print a 4×6 rectangle of stars.

```python
for i in range(1, 5):
    print()
    for j in range(1, 7):
        print("*", end="")
```

### 2: Print multiplication tables from 1–5.

```python
number = int(input("Enter a number: "))
for i in range(1, 11):
    print(f"{number} x {i} = {number * i}")
```

### 3: Diamond-like half pattern.

```python
rows = int(input("Enter Half Diamond Pattern Rows = "))
print("Half Diamond Star Pattern")
for i in range(rows):
    for j in range(0, i + 1):
        print('*', end = '')
    print()
for i in range(1, rows):
    for j in range(i, rows):
        print('*', end = '')
    print()
```

### 4: Print A, AB, ABC, ABCD, ABCDE.

```python
alphabet = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
rows = int(input("Enter the number of rows: "))
for i in range(rows):
    for j in range(i + 1):
        print(alphabet[j], end=' ')
    print()
```

## **Break & Continue**

### 1: Stop at 13 while printing 1–20.

```python
for i in range(1, 20):
    if i == 14:
        break
    print(i, end=" ")
```

### 2: Skip multiples of 3 from 1–20.

```python
for i in range(1, 20):
    if i%3== 0:
        continue
    print(i, end=" ")
```

### 3: Print a 5×5 grid skipping the centre

```python
for i  in range(1, 6):              # outter loop for rows
    for j in range(1, 6):           # inner loop for columns
        if i == 3 and j ==3:        # condition to check if the current row and column is 3,3
            print("X", end=" ")     # print X instead of * if the condition is true
            continue                # skip the rest of the inner loop and move to the next iteration
        print("*", end=" ")         # print * if the condition is false
    print()                         # move to the next line after each row is printed
```
