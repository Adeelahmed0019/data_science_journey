### **1. Basic Loops (for and while)Python features two primary loop types:** 

#### **for loop**: Iterates over a sequence (such as a list, tuple, string, or range()).

```python
# For Loop Example
x = 6
for i in range(x):
    print(f"x: {i}")
```

## 2: **Range()** 
#### It is predefine python funcion that is mostly used for printing numbers in specfic `range`
#### It takes three values [start: stop: step ]
### `start` = first/initial value from where `range()` function `start` to print
### `stop`  = End(n-1) value where the `range()` function `stoped`
### `step`  = gap between the numbers 
#### By default it will `stop` just one value/iteration before stop value.
#### If you don't mention `step` value it automatically consider 1 as a `step` value. 
#### If you just put only `single` digit value and you did't mention `stop` and `step` value in `range()` function, it by default kprint value from zero

```python
print(range(6))
```

#### 3: **while loop**: Executes continuously as long as a specified condition remains True.

```python
# While Loop Example
count = 0
while count < 3:
    print(f"While count: {count}")
    count += 1
```

## 4: **break**
### Exits the loop entirely.
### Jumps to the first line of code after the loop block.

```python
# Example of break
print("Testing break:")
for num in range(1, 6):
    if num == 3:
        break  # Stops the loop completely when num is 3
    print(num)  # Outputs: 1, 2
```

## 5: **continue** 
### Skips the rest of the current iteration.
### Jumps back to the top of the loop for the next cycle.

```python
#Example of continue
print("Testing continue:")
for num in range(1, 6):
    if num == 3:
        continue  # Skips printing 3 and moves to 4
    print(num)  # Outputs: 1, 2, 4, 5
```

## 6: **Nested Loops**
### A nested loop is a `loop inside another loop`. 
### The inner loop executes completely for every individual iteration of the outer loop.
### outter loop = Rows
### inner loop  = coloumns

```python
# Nested loop example # reference: https://www.w3schools.com/python/python_for_loops.asp
for outer in ["A", "B"]:            # Outer loop iterates over the list ["A", "B"]
    for inner in [1, 2]:            # Inner loop iterates over the list [1, 2]
        print(outer, inner)         # Outputs: A 1, A 2, B 1, B 2
```

## 7: **break and continue in Nested Loops**
### Does break stop both loops?
### answer: NO

```python
# reference: https://www.w3schools.com/python/python_for_loops.asp
for outer in range(1, 4):                       # Outer loop iterates over the range 1 to 3
    for inner in range(1, 4):                   # Inner loop iterates over the range 1 to 3
        if inner == 2:                          # When inner is 2, break out of the inner loop
            break                               # This only breaks out of the inner loop
        print(f"Outer: {outer}, Inner: {inner}")
```

## **How to break out of both loops simultaneously**

```python
for i in range(3):
    for j in range(3):
        print(f"Outer: {i}, Inner: {j}") 
        if i == 2 and j == 2:
            break # Breaks inner loop
```
