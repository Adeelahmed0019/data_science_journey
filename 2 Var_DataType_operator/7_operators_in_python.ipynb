# BMI: body mass index
## formula = weight multiply by square of the height

```python
weight = int(input("Enter your 'weight' as 'integer' number"))
print("Your 'weight' is: -->",weight)

height = float(input("Enter your 'height'"))
print("Your 'height' is: -->",height)

# # Apply formula:
formula = weight/(height**2)
print(formula)
print("Your calculated value is: -->", round(formula, 2))
```

```python
# 1. Take weight input as an integer
weight = int(input("Enter your 'weight' as 'integer' number: "))
print("Your 'weight' is: -->", weight)
# 2. Take height input as a float
height = float(input("Enter your 'height': "))
print("Your 'height' is: -->", height)


# 3. Apply the correct formula (Height squared)
formula = weight / (height ** 2)
# Round the output to 2 decimal places for cleaner viewing
```
