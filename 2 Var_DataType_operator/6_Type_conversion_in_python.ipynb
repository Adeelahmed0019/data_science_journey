# **Type Conversion** 
#### You can convert from one type to another with the int(), float(), and complex() methods: 
Note: You cannot convert complex numbers into another number type. 

## Convert from one type to another

```python
# Example

# int 
x = 1     
print("The value of 'x' before CASTING is :" ,x)

# float
y = 2.8   
print("The value of 'y' before CASTING is :" ,y)

# Complex 
z = 1j    
print("The value of 'z' before CASTING is :" ,z)

print("*"*50) 

#convert from int to float 
a = float(x)
print("The value of 'x' after CASTING is :" ,a,"and its datatype is :", type(a) ) 
   
#convert from float to int
b = int(y)      
print("The value of 'y' after CASTING is :" ,b,"and its datatype is :", type(b) ) 

#convert from int to complex
c = complex(x) 
print("The value of 'z' after CASTING is :" ,c,"and its datatype is :", type(c) ) 
 
 
```

### **First level**

```python
first = int(input("Enter first value"))
print("The first value is: ",first) 
second = int(input("Enter second value"))
print("The second value is: ",second) 

add = first + second
print("The addition of", first, "and" ,second, "is: ",add)
```

# **Variation** of First level

```python
first = input("Enter first value")
print("The first value is: ",first) 
second = input("Enter second value")
print("The second value is: ",second) 

add = int(first) + int(second)
print("The addition of", first, "and" ,second, "is: ",add)
```

### Take the input from `User` as any `two digit` numbers like *12* and the add them as 1+2 = 3
### Note: Take a single `input`
### **Second level**: only one `input` and any `two` digit number

```python
number = input("Enter any two digit number")
print("The first value is: ",number)
add = int(number[0]) + int(number[1])
print(add)

 
```

### **Third level**: Single `input` with `multi-digit` value

```python
number = input("Enter any multi-digit value: ")
print("The entered value is:", number)

# Initialize a counter
total_sum = 0

# Loop through every digit in the string
for digit in number:
    if digit.isdigit():  # Security check to skip negative signs or accidental letters
        total_sum += int(digit)

print("The sum of all digits is:", total_sum)
```

### **Advance level:**

```python
# Accept comma-separated numbers from the user
number = input("Enter numbers separated by commas: ")
print("The entered value is:", number)

# Split the string by commas into a list of strings
number_list = number.split(',')

total_sum = 0

# Loop through each item in the split list
for item in number_list:
    cleaned_item = item.strip() # Removes any accidental spaces
    if cleaned_item.isdigit():  # Ensures it is a valid number
        total_sum += int(cleaned_item)

print("The sum of all numbers is:", total_sum)
```
