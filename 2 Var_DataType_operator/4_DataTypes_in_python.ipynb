# **Python Data Types**
### Built-in Data Types 
In programming, `data type` is an important concept. 
Variables can store data of different types, and different types can do different things.
### ***Python has the following data types built-in by default, in these categories***


#### 1 Text Type: str 
#### Numeric Types: int, float, complex 
#### Sequence Types: list, tuple, range  
#### Mapping Type: dict 

#### Set Types: set, frozenset
#### Boolean Type: bool
#### Binary Types: bytes, bytearray, memoryview
#### None Type: NoneType


## ***Getting the Data Type***
#### You can get the data type of any object by using the `type()` function: 

```python
# Print the data type of the variable x: 
x = 5 
print("The value of 'x' is : ==> " , x)
print("The dataType of 'x' is : ==> ",type(x)) 
```

### Setting the Data Type 
In Python, the data type is set when you assign a value to a variable
#### ***We divided `data-types` into `three` categories***

# The first category is:

```python
# Example
print("*"*68)
print("First category of datatypes: ")
print("Here we did not specify the 'data-types' python itself set them\n")
x = "Hello World" 
print("(1).",x,"\t The datatype of 'x' is:",type(x))
x = 20  
print("(2).",x,"\t\t The datatype of 'x' is:",type(x))
x = 20.5
print("(2).",x,"\t\t The datatype of 'x' is:",type(x))
x = 1j 
print("(4).",x,"\t\t The datatype of 'x' is:",type(x))
x = range(5) 
print("(5).",x,"\t The datatype of 'x' is:",type(x))
```

# The Second category is:

```python
print("*"*85)
print("Second category of datatypes: ")
print("Here we did not specify the 'data-types' python itself set them\n")
x = ["apple", "banana", "cherry"] 
print("(6).",x,"\t The datatype of 'x' is:",type(x))
x = ("apple", "banana", "cherry") 
print("(7).",x,"\t The datatype of 'x' is:",type(x))
x = {"name" : "John", "age" : 36}
print("(8).",x,"\t The datatype of 'x' is:",type(x))
x = {"apple", "banana", "cherry"}
print("(9).",x,"\t The datatype of 'x' is:",type(x))
```

# The Third category is:

```python
print("*"*95)
print("Third category of datatypes: ")
print("Here we did not specify the 'data-types' python itself set them\n")
x = frozenset({"apple", "banana", "cherry"}) 
print("(10).",x,"\t The datatype of 'x' is:",type(x))
x = True
print("(11).",x,"\t\t\t\t\t The datatype of 'x' is:",type(x))
x = b"Hello" 
print("(12).",x,"\t\t\t\t\t The datatype of 'x' is:",type(x))
x = bytearray(5)  
print("(13).",x,"\t The datatype of 'x' is:",type(x))
x = memoryview(bytes(5)) 
print("(14).",x,"\t\t The datatype of 'x' is:",type(x))
x = None  
print("(15).",x,"\t\t\t\t\t The datatype of 'x' is:",type(x))
```

### **Setting the Specific Data Type**
If you want to specify the data type, you can use the following constructor functions: 
# The First category is:

```python
# Example
print("*"*60)
print("First category of datatypes: ")
print("Here we specify the 'data-type manually\n")
x = str("Hello World") 
print("(1).",x,"\t The datatype of 'x' is:",type(x))
x = int(20)  
print("(2).",x,"\t\t The datatype of 'x' is:",type(x))
x = float(20.5)
print("(2).",x,"\t\t The datatype of 'x' is:",type(x))
x = complex(1j) 
print("(4).",x,"\t\t The datatype of 'x' is:",type(x))
x = range(5) 
print("(5).",x,"\t The datatype of 'x' is:",type(x))
print("*"*60)
```
