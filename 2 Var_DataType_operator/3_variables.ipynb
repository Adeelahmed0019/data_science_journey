# 1: **`Variables in Python`** 

### `Variables` are `containers` for store the data. 
#### In Python, variables are created when you assign a value to it: 
#### Python has no command for declaring a variable.


## 2: **Creating Variables**

```python
# Example 
x = 1
y = "python" 
print(x) 
print(y) 
```

#### Variables `do not` need to be `declared` with any particular `type`, and can even `change type` after they have been sat.

```python
# Example 
a = 2                                   # a is of type int 
print("Here 'a' has value of :",a)      # it will print 2
print("The type of 'a' is : ", type(a)) # it will print the type of 'a'

print("-"*50)                           # just to print a line of 50 dashes")

a = "programming"                       # a is now of type str 
print("Here 'a' has value of :",a)      # it will print "programming"
print("The type of 'a' is : ", type(a)) # it will print the type of 'a'
```

## 3: **Casting**
#### If you want to specify the data type of a variable, this can be done with casting. 

```python
# Example 
a = str(2)    
print("The value of 'a' is :",a)                  # x will be '2' 
print("The data_type of 'a' is :", type(a)) 
 
b = int(2)    
print("The value of 'b' is :",b)                  # y will be 2 
print("The data_type of 'b' is :", type(b)) 

c = float(2) 
print("The value of 'c' is :",c)                  # z will be 2.0 
print("The data_type of 'a' is :", type(c)) 
```

# 5: **Case-Sensitive**
#### Python Variables names are case-sensitive. 

```python
# Example 
# This will create two variables: 
a = 4 
print(a)

A = "Good" 
print(A)
# Note: 'A' will not overwrite 'a' because Python is case-sensitive. So, 'a' and 'A' are two different variables. 

# checking if 'a' and 'A' are the same or not.
print(a == A)
```

# 7: **There are several techniques you can use to make them more readable:** 
### • Camel Case 
### • Pascal Case 
### • Snake Case 

# `Camel Case` 
#### Each word, `except the first`, starts with a `capital letter` 


```python
# Example 1
myVariableName = "Jonson"
print(myVariableName)

# Example 2
myAge = 25
print(myAge)
```

# `Pascal Case`
### Each word starts with a `capital letter` 


```python
# Example 1
MyVariableName = "Jonson"
print(MyVariableName)

# Example 2
MyAge = 25
print(MyAge)
```

# `Snake Case` 
### Each word is `separated` by an `underscore` character


```python
# Example 1
my_variable_name = "Jonson" 
print(my_variable_name)

# Example 2
my_age = 25
print(my_age)
```

# 8: **Python Variables Assign Multiple Values** 
## multiple Values to Multiple Variables
##### Python allows you to assign values to multiple variables in one line
##### Note: Make sure the number of variables matches the number of values, or else you will get an error(ValueError).  

```python
# Example   
a, b, c = "First", "Second", "Third" # Here we assinging Multiple valuse to multiple variables
print(a) 
print(b) 
print(c)
```

## `One Value to Multiple Variables` 
#### You can assign the same value to `multiple` variables in one line

```python
# Example         
a = b = c = "First" 
print(a) 
print(b) 
print(c) 
```

# `packing the values`
#### ValuesPacking happens when you bunch multiple independent values together into a single container object. 
#### How it looks: Python automatically packs 'comma-separated' values into a tuple without requiring parentheses.

```python
# Example
fruits = "apple", "banana", "cherry" 
a, b, c = fruits 
print("Here we print 'Packed' values",fruits)
print(type(fruits))
```

# `Unpack a Collection`
#### If you have a collection of values in a list, tuple etc. Python allows you to extract the values into variables **This is called unpacking**
#### Note: if the number of `variables` and number of `items` does't matched there will be an error(ValueError)

```python
print(fruits)
a, b, c, = fruits
print("unpacked values: ",a, type(a)) 
print("unpacked values: ",b, type(b)) 
print("unpacked values: ",c, type(c))
```

# 9: **Python - Global Variables** 
### Global Variables that are created outside of a function (Block of code) are known as global variables. 
#### Note: `Global variables can be used by everyone, both inside of functions and outside`
Create a variable outside of a function, and use it inside the function 


```python
# Example 
x = "awesome" 
def myfunc(): 
  print("Python is "+ x) 
myfunc()
```

If you create a variable with the `same name inside a function`, this variable will be `local`, and can 
only be used `inside the function`. 
The `global variable` with the `same name` will remain as it was, global and with the original value. 
#### **Create a variable inside a function, with the same name as the global variable**

```python
x = "awesome"                       # Global variable bc it is created "outside" of the Function
def myfunc():                       # Function defined here
  x = "fantastic"                   # Local variable bc it is created "inside" the function
  print("Python is " + x)           # Print "Local" variable
myfunc()                            # Here we're "calling" the Function
print("Python is " + x)             # Here we're printing "Global" variable
```

### **The global Keyword** 
##### Normally, when you create a variable `inside` a function, that variable is `local`, and can only be used inside that function.
### `To create a global variable inside a function, you can use the global keyword`
If you use the `global` keyword, the variable belongs to the global scope: 
#### Note: Once you declared a variable name with `Global` keyword, then you can not `use` that variable name again, if you did you will get an error like `SyntaxError: name 'x' is assigned to before global declaration`


```python
#Example
 
f = "Good"          # Just Global varible

def myfunc():     
  y = "Awesome"       # local variable inside the function
  
  global x            # Global variable with "global-Keyword"
  x = "fantastic" 
  print("Global variable 'x' called inside the function",x)       
  print("Local variable 'y' form function called inside the function",y)       
myfunc() 
print("Global variable 'x' called outside the function",f)  
print()     

# Note #1: There two types of "Global" variable declaration. 
# 1st: You declare a global varible 'Outside' of the function
# 2nd: you declare a global variable 'inside' the function with "global" keyword

# Observation says:
# 2nd one has higher priority than 1st one


```
