# # A: **List Methods**

### 1: **sort()**
#### Sort the list in ascending order

```python
nums = [5,2,9,1,7]
print("Unsorted list => ",nums)
nums.sort()
print("Sorted list => ",nums)  # Output: [1, 2, 5, 7, 9]

# sort in reverse order
nums.sort(reverse=True)
print("Sorted list in reverse order => ",nums)  # Output: [9, 7, 5, 2, 1]
```

### 2: **sorted()**
### return the sorted list but `original` remains untuched


```python
data = [3,1,2]
print("origianl/Unsorted list => ",data)
new_data = sorted(data)  
print("Sorted list => ",new_data)  # Output: [1, 2, 3]
print(data)
```

### 3: **reverse()**
#### Reverse the list itmes

```python
r_data = [3,1,2]
print("origianl/Unsorted list => ",r_data)
print("origianl/Unsorted list => ",r_data)
r_data.reverse()
print("Reversed list => ",r_data)  # Output: [2, 1, 3]

# origianl list =>  [1, 2, 3]
original = [1, 2, 3]
flipped = original[::-1]
print("Reversed list => ",flipped)  # Output: [3, 2
```

### 4: **count()**
#### Return the number of times x appear

```python
fruits = ['apple', 'banana', 'cherry', 'apple']
print(fruits.count('apple'))  # Output: 2
```

### 5: **index()**
#### Return the position of first matching x

```python
fruits = ['apple', 'banana', 'cherry', 'apple']
print(fruits.index('apple'))  # Output: 1 [first match]
```

### 6: **extend()**
#### Add all items of the list_2 at the end

```python
a = [1, 2, 3]
print("List `a` is => ",a)  # Output: [1, 2, 3]

b = [4, 5, 6]
print("List `b` is => ",b)  # Output: [4, 5, 6]

a.extend(b)
print("extended list is => ",a)  # Output: [1, 2, 3, 4, 5, 6]
```

# B: **`List comprehension`**

### List comprehension is a concise, elegant way to create a new list from an existing iterable (like a list, tuple, string, or range) using a single line of code.

# `The Basic Syntax`
## **new_list = [expression for item in iterable if condition]**
#### • `expression`: The operation or transformation you want to apply to each item.
#### • `item`: The current element being processed from the iterable.
#### • `iterable`: The collection you are looping through (e.g., range(), a list of words, etc.).
#### • `if condition (Optional)`: A filter that only includes items that evaluate to True.

# **Standard Loop vs. List Comprehension**
## The Traditional for Loop Approach

```python
numbers = [1, 2, 3, 4, 5]
squares = []

for num in numbers:
    squares.append(num ** 2)

print(squares)  # Output: [1, 4, 9, 16, 25]
```

## *The List Comprehension Equivalent*

```python
numbers = [1, 2, 3, 4, 5]

squares = [num ** 2 for num in numbers]

print(squares)  # Output: [1, 4, 9, 16, 25]
```

# **Adding Logic (Filtering & Modifications)**

## `'If' as a Filter`
#### if at the end (Filter): It acts as a gatekeeper. It decides whether or not an item gets into the new list.

```python
# Structure: [expression for item in iterable if condition]
numbers = [1, 2, 3, 4, 5, 6]

# Only keep numbers greater than 3
filtered = [num for num in numbers if num > 3]

print(filtered)  
# Output: [4, 5, 6] (Notice the list shrunk!)
```

## `'if' as a statement`
#### if at the beginning (Statement/Expression): It acts as a transformer. It modifies the item depending on a condition, ensuring every item passes through but changes its shape.

```python
# Structure: [true_expr if condition else false_expr for item in iterable]
numbers = [1, 2, 3, 4, 5, 6]

# Replace numbers: if it's greater than 3, keep it. Otherwise, make it 0.
transformed = [num if num > 3 else 0 for num in numbers]

print(transformed)  
# Output: [0, 0, 0, 4, 5, 6] (Same length, but values are modified)
```

## `Can you combine both?`

#### Yes! You can use an if-else statement at the beginning to transform data, and an if filter at the end to drop data simultaneously.

```python
numbers = [1, 2, 3, 4, 5, 6]

# 1. Filter: Only look at numbers greater than 2 (drops 1 and 2)
# 2. Transform: Label the remaining numbers as 'Even' or 'Odd'
result = ["Even" if num % 2 == 0 else "Odd" for num in numbers if num > 2]

print(result)  
# Output: ['Odd', 'Even', 'Odd', 'Even']
```
